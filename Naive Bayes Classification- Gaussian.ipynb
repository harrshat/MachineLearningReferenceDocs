```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Used to solve classification - binary and multi class classification
```

```python
from sklearn.datasets import load_iris
```

```python
from sklearn.model_selection import train_test_split
```

```python
X,y=load_iris(return_X_y=True)
#return_X_y - return dependent and independent features
```

```python
X.shape
```

```python
y.shape
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=10)
```

https://scikit-learn.org/0.22/modules/generated/sklearn.naive_bayes.GaussianNB.html

```python
from sklearn.naive_bayes import GaussianNB
```

```python
model=GaussianNB()
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
print("Accuracy Score:\n")
print(accuracy_score(y_test,y_pred))
print("Classification report:\n")
print(classification_report(y_test,y_pred))
print("Confusion Matric:\n")
print(confusion_matrix(y_test,y_pred))
```
