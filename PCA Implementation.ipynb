## Aim:  Reduce the number of features from 30 available features using PCA

```python
X = pd.DataFrame(load_breast_cancer()['data'],columns=load_breast_cancer()['feature_names'])
X.head()
```

```python
y=pd.DataFrame(load_breast_cancer()['target'],columns=['output'])
y.head()
```

```python
y['output'].value_counts()
# 0 - Malignant
# 1 - Benign
```

```python
# Standardization
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
scaler.fit(X)
```

```python
scaled_data = scaler.transform(X)
```

```python
scaled_data # Mean - 0 , Std deviation 1
```

```python
# Applying PCA
from sklearn.decomposition import PCA
```

```python
pca=PCA(n_components=2)
'''
(n_components: Any | None = None, *, copy: bool = True, whiten: bool = False,
svd_solver: str = "auto", tol: float = 0, iterated_power: str = "auto", n_oversamples: int = 10,
power_iteration_normalizer: str = "auto", random_state: Any | None = None) -> PCA
'''
#n_components - How many features are wanted from existing features

data_pca = pca.fit_transform(scaled_data)
```

```python
pca_dataframe = pd.DataFrame(data_pca,columns=['feature1','feature2'])
pca_dataframe.head(5)
```

```python
pca.explained_variance_
# feature1, feature2
```

```python
plt.figure(figsize=(10,10))
plt.scatter(pca_dataframe['feature1'],pca_dataframe['feature2'],c=y['output'])
plt.xlabel("First Principal Component")
plt.ylabel("Second Principal Component")
```
