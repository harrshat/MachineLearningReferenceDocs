```python
# sex, smoker and time are binary categories - feature encoding (label and one hot)
# Label - replace with 1 and 0 - 2 categories only
# One Hot Encoding [0 0 1 0] etc.
```

```python
#Independent and dependent features
X=df[['tip', 'sex', 'smoker', 'day', 'time', 'size']]
y=df['total_bill']
```

```python
X.shape
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
X_train.shape
```

```python
y_train.shape
```

```python
X_test.shape
```

```python
y_test.shape
```

```python
# Feature Encoding - Label Encoding and One Hot Encoding
from sklearn.preprocessing import LabelEncoder
le1=LabelEncoder()
le2=LabelEncoder()
le3=LabelEncoder()
```

```python
X_train.head(3)
```

```python
X_train['sex']=le1.fit_transform(X_train['sex'])
X_train['smoker']=le2.fit_transform(X_train['smoker'])
X_train['time']=le3.fit_transform(X_train['time'])
```

```python
X_test['sex']=le1.transform(X_test['sex'])
X_test['smoker']=le2.transform(X_test['smoker'])
X_test['time']=le3.transform(X_test['time'])
```

```python
X_train.head(5)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
ct= ColumnTransformer(transformers=[('onehot',OneHotEncoder(drop='first'),[3])],remainder='passthrough')
```

```python
ct
```

```python
X_train=ct.fit_transform(X_train)
```

```python
X_train
```

```python
X_test=ct.transform(X_test)
```

```python
X_test[0:5]
```

# Support Vector Regressor

```python
from sklearn.svm import SVR
```

```python
svr=SVR()
```

```python
svr.fit(X_train,y_train)
```

```python
y_pred=svr.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
```

```python
print(r2_score(y_test,y_pred))
```

```python
print(mean_absolute_error(y_test,y_pred))
```
